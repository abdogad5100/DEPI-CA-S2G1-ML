###  To fix this issue You will fill these values with median for numeric and mode for categorical

```python
# trying to fill missing values in my datset using for loop 
numeric_columns = ["member_birth_year"]


numeric_columns = ["member_birth_year"]
categorical_columns = ["member_gender"]

for col in numeric_columns:
    df[col] = df[col].fillna(df[col].median())

for col in categorical_columns:
    df[col] = df[col].fillna(df[col].mode()[0])



```

```python
## create a function to know the average users ages
def calculate_average_age(df: pd.DataFrame) -> float:
    """
    Calculate the average age of users.

    Parameters
    ----------
    df : pd.DataFrame
        The input DataFrame containing the 'user_age' column.

    Returns
    -------
    float
        The average age of users.
    """
    return df["user_age"].mean()
```

```python
# Make visualization to easy get insight

plt.figure(figsize=(10, 5), dpi=100)

ax = sns.countplot(data=df, x='bike_share_for_all_trip')

total = len(df)
for p in ax.patches:
    percentage = f'{100 * p.get_height() / total:.1f}%'
    ax.annotate(percentage, (p.get_x() + p.get_width() / 2, p.get_height()),
                ha='center', va='bottom')

plt.xlabel('Bike Share for All Trip')
plt.ylabel('Frequency')
plt.title("Distribution of Bike Share Users on All Trips")

plt.show()
```


# <a id="13">  13 ) . What is the  Correlation between each variable in the dataset?</a>

```python
# Make visualization to easy get insight
plt.figure(figsize=(12, 8))

correlation = df.corr(numeric_only=True)

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Between Variables')
plt.show()
```

## Insight


- `member_birth_year` and `user_age` show a **perfect negative correlation (-1.00)**, which makes sense because age was calculated directly from birth year (`current_year - birth_year`).
- `start_station_latitude` and `end_station_latitude` show a **very strong positive correlation (0.99)**, and similarly `start_station_longitude` and `end_station_longitude` show a **very strong positive correlation (0.99)**, indicating that most trips start and end at geographically close locations.
- `start_station_latitude` and `start_station_longitude` show a **strong negative correlation (-0.68)**, and the same pattern appears between `end_station_latitude` and `end_station_longitude` (-0.68), reflecting the geographic layout of the bike stations across the city.
- `start_station_id` and `end_station_id` show a **moderate positive correlation** with the longitude variables (around 0.46–0.48), suggesting station IDs are somewhat related to their geographic location.
- `bike_id` shows **very weak or no correlation** with almost all other variables (values close to 0), confirming that bike assignment is random and unrelated to trip duration, location, or user demographics.
- Trip duration variables (`duration_sec`, `duration_min`, `duration_hour`) show **no meaningful correlation** with any other variable in the dataset, including station location, user age, or bike ID — all values are close to 0.
