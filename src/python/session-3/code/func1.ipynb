```python
def info_user( age: int | None = None , name: str | None = None
) -> list:
    '''
    this function will return the name and age of the user
    Args:
        name (str): the name of the user
        age (int): the age of the user
    Returns:
        list: a list containing the user's name and age
    '''
    print(f"my name is {name}, my age is {age} years old.")
    return [name, age]
```

```python
def train_model(model_name ,*dataset, **hyperparameters):
    print(f"model name is {model_name}")
    for data in dataset:
        print(f"*", data)
    print(f"hyperparameters are {hyperparameters}")
    for key, value in hyperparameters.items():
        print(f"key is {key} and value is {value}")
```

```python
train_model("fraud_detection","train_model.csv","validation.csv",
            "test.csv",learning_rate=0.001,
              batch_size=64,
                epochs=100,
                optimizer="Adam",
                early_stopping=True )
```
