## Task-3: 
### Encoded Message: ##$$$@!yalpstcejorp EPUVT****9887 
### 1. Extract the core part of the message: "yalpstcejorp EPUVT". 

```python
encoded_message = "##$$$@!yalpstcejorp EPUVT****9887"
new_encoded_message = encoded_message.split("!")[1].rsplit("*")[0]
new_encoded_message
```

### 2. Reverse the first word: "yalpstcejorp" becomes "projectplay".

```python
first_word = new_encoded_message.split(" ")[0]
first_word = first_word[::-1]
first_word
```

### 3. Replace shifted vowels in the second word: 
### "EPUVT": Replace E->A, U->O to get "APOVT". 

```python
second_word = new_encoded_message.split(" ")[1]
second_word = second_word.lower()
print(f"Second word: {second_word}")
vowels = "aeiou"
if("e" in second_word or "u" in second_word):
    second_word = second_word.replace("e", "a").replace("u", "o")
       
second_word = second_word.upper()
print(f"Second word: {second_word}")
```

### 4. Final decoded message: "projectplay APTOV". 

```python
final_message = first_word + " " + second_word
print(f"Final message: {final_message}")
```
