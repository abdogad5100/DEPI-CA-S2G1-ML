## Task-3: 
### Encoded Message:  &&&**$gnirtS PLIO!!@1234 
### 1. Extract the core part of the message: "gnirtS PLIO". 


```python
encoded_message = "&&&**$gnirtS PLIO!!@1234"
new_encoded_message = encoded_message.strip("&*$").rsplit("!")[0]
new_encoded_message
```

### 2. Reverse the first word: "gnirtS" becomes "String". 

```python
first_word = new_encoded_message.split(" ")[0]
first_word = first_word[::-1]
first_word
```

### 3. Replace shifted vowels in the second word: 
### "PLIO": Replace I->E and O->U to get "PLEU". 

```python
second_word = new_encoded_message.split(" ")[1]
second_word = second_word.lower()
print(f"Second word: {second_word}")
vowels = "aeiou"
if("i" in second_word and "o" in second_word):
    second_word = second_word.replace("i", "e").replace("o", "u")
       
second_word = second_word.upper()
print(f"Second word: {second_word}")
```

#### 4. Final decoded message: "Welcome PGTQ". 


```python
final_message = first_word + " " + second_word
print(f"Final message: {final_message}")
```
