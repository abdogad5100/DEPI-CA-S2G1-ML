## Encoded Message: 
### ###!!@mocleW EPGTQ!!!6789
##### 1. Extract the core part of the message: "mocleW EPGTQ".

```python
encoded_message = "###!!@emocleW EPGTQ!!!6789"
new_encoded_message = encoded_message.strip("#!@").split("!")[0]
new_encoded_message
```

#### 2. Reverse the first word: "mocleW" becomes "Welcome". 


```python
first_word = new_encoded_message.split(" ")[0]
first_word = first_word[::-1]
first_word
```

#### 3. Replace shifted vowels in the second word:
 "EPGTQ": No vowels to change. 

```python
second_word = new_encoded_message.split(" ")[1]
second_word = second_word.lower()
print(f"Second word: {second_word}")
vowels = "aeiou"
for vowel in vowels:
    if vowel in second_word:
        second_word = second_word.replace(vowel, "")
       
second_word = second_word.upper()
print(f"Second word: {second_word}")
```

#### 4. Final decoded message: "Welcome PGTQ". 


```python
final_message = first_word + " " + second_word
print(f"Final message: {final_message}")
```
