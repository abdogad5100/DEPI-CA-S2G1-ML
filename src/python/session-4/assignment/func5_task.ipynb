```python
def perfectNums(num1: int, num2: int) -> None:
    """
    Find and print all perfect numbers within a given range.

    A perfect number is a number that is equal to the sum
    of its proper divisors, excluding the number itself.

    Args:
        num1 (int): The starting number of the range.
        num2 (int): The ending number of the range.

    Returns:
        None: This function only prints the perfect numbers.

    Example:
        perfectNums(0, 100)

        Output:
            0
            6
            28
    """

    for i in range(num1, num2):

        divisor_sum = 0

        for divisor in range(1, i):
            if i % divisor == 0:
                divisor_sum += divisor

        if divisor_sum == i:
            print(i)
```

```python
perfectNums(0, 100)
```
