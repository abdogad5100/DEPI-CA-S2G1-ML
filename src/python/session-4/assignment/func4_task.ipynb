```python
def decToBin(num: int) -> str:
    """
    Convert a decimal number to its binary representation.

    Args:
        num (int): The decimal number to convert.

    Returns:
        str: The binary representation of the given decimal number.

    Example:
        decToBin(11)

        Output:
            "1011"
    """

    binary = ""

    while num > 0:
        if num % 2 == 0:
            binary += "0"
        else:
            binary += "1"

        num = num // 2

    return binary[::-1]
```

```python
result = decToBin(11)
result
```
