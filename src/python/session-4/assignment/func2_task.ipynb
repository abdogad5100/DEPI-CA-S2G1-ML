### 2. Write a program to print twin primes less than 1000. If two 
### consecutive odd numbers are both prime then they are known as 
### twin primes 

```python
def is_prime(num: int) -> bool:
    """
    Check whether a given number is a prime number.

    Args:
        num (int): The number to check.

    Returns:
        bool: True if the number is prime, otherwise False.

    Example:
        is_prime(7)

        Output:
            True
    """        
    
    if num <= 1:
        return False

    for divisor in range(2, num):
        if num % divisor == 0:
            return False

    return True
```

```python
is_prime(3)
```

```python
def print_twin_primes(limit: int) -> None:
    """
    Find and print all twin prime pairs below a given limit.

    Args:
        limit (int): The upper limit for finding twin prime pairs.

    Returns:
        None: This function only prints the twin prime pairs.

    Example:
        print_twin_primes(10)

        Output:
            Twin primes:
            3 and 5
            5 and 7
    """

    print("Twin primes:")

    for number in range(3, limit):
        if is_prime(number) and is_prime(number + 2):
            print(f"{number} and {number + 2}")
```

```python
print_twin_primes(1000)
```
