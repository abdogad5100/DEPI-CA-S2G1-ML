```python
def prime_factors(num: int) -> list:
    """
    Find the prime factors of a given number.

    Args:
        num (int): The number to find its prime factors.

    Returns:
        list: A list containing the prime factors of the number.

    Example:
        prime_factors(56)

    Returns:
        [2, 2, 2, 7]
    """

    divisor = 2
    prime_factors_list = []

    while num > 1:
        if num % divisor == 0:
            prime_factors_list.append(divisor)
            num = num // divisor
        else:
            divisor += 1

    return prime_factors_list
```

```python
result = prime_factors(56)
result
```
