### 1. Write a function that inputs a number and prints the multiplication table of that number 

```python
def multi_table(number, multiplication_limit) -> None:
    """
    Generate and print a multiplication table for a given number.

    Args:
        number (int): The number to generate the multiplication table for.
        multiplication_limit (int): The maximum multiplier.

    Returns:
        None: This function only prints the multiplication table.

    Example:
        multi_table(1, 3)

        Output:
            1 x 1 = 1
            1 x 2 = 2
            1 x 3 = 3             
    """

    for multiplier in range(multiplication_limit):
        multiplication_result = (multiplier + 1) * number

        print(f"{number} x {multiplier + 1} = {multiplication_result}")

```

```python
multi_table(9,12)    
```
