```python
class Factorial:
    """
    A class used to calculate the factorial of a number using recursion.

    Attributes:
        num (int): The number whose factorial will be calculated.
    """

    def __init__(self, num: int):
        """
        Initialize the Factorial object with a number.

        Args:
            num (int): The number whose factorial will be calculated.
        """
        self.num = num

    def calculate_factorial(self, num: int = None) -> int:
        """
        Calculate the factorial of a number using recursion.

        Args:
            num (int, optional): The number used in the recursive calculation.
                If not provided, the object's number is used.

        Returns:
            int: The factorial of the number.

        Example:
            Factorial(5).calculate_factorial() returns 120.
        """
        if num is None:
            num = self.num

        if num == 0:
            return 1

        return num * self.calculate_factorial(num - 1)


# Create an object
factorial_object = Factorial(5)

# Call the main method
result = factorial_object.calculate_factorial()

# Print the result
print(result)
```

```python
class PrimeChecker:
    """
    A class used to check whether a number is prime or not.

    Attributes:
        num (int): The number to be checked.
    """

    def __init__(self, num: int):
        """
        Initialize the PrimeChecker object with a number.

        Args:
            num (int): The number to be checked.
        """
        self.num = num

    def is_prime(self) -> bool:
        """
        Check whether the stored number is prime or not.

        Returns:
            bool: True if the number is prime, otherwise False.

        Example:
            PrimeChecker(17).is_prime() returns True.
        """
        if self.num < 2:
            return False

        for i in range(2, int(self.num ** 0.5) + 1):
            if self.num % i == 0:
                return False

        return True


# Create an object
prime_checker = PrimeChecker(17)

# Call the method
result = prime_checker.is_prime()

# Print the result
print(result)
```

```python
class CommonDivisor:
    """
    A class used to find the common divisors of two numbers.

    Attributes:
        num1 (int): The first number.
        num2 (int): The second number.
    """

    def __init__(self, num1: int, num2: int):
        """
        Initialize the CommonDivisor object with two numbers.

        Args:
            num1 (int): The first number.
            num2 (int): The second number.
        """
        self.num1 = num1
        self.num2 = num2

    def find_common_divisors(self) -> list[int]:
        """
        Find all common divisors of the two stored numbers.

        Returns:
            list[int]: A list containing all common divisors.

        Example:
            CommonDivisor(12, 18).find_common_divisors()
            returns [1, 2, 3, 6].
        """
        limit = min(self.num1, self.num2)
        divs = []

        for divisor in range(1, limit + 1):
            if self.num1 % divisor == 0 and self.num2 % divisor == 0:
                divs.append(divisor)

        return divs


# Create an object
common_divisor = CommonDivisor(12, 18)

# Call the main method
result = common_divisor.find_common_divisors()

# Print the result
print(result)
```

```python
class TextReverser:
    """
    A class used to reverse the order of words in a text.

    Attributes:
        text (str): The text whose words will be reversed.
    """

    def __init__(self, text: str):
        """
        Initialize the TextReverser object with a text.

        Args:
            text (str): The text to be reversed.
        """
        self.text = text

    def reverse_text(self) -> list[str]:
        """
        Reverse the order of words in the stored text.

        Returns:
            list[str]: A list containing the words in reverse order.

        Example:
            TextReverser("Hello World Python").reverse_text()
            returns ["Python", "World", "Hello"].
        """
        words = self.text.split()
        reversed_words = words[::-1]

        return reversed_words


# Create an object
text_input = input("Enter the text Uwant to reverse:")
text_reverser = TextReverser(text_input)

# Call the main method
result = text_reverser.reverse_text()

# Print the result
print(result)
```

```python
import random


class PasswordGenerator:
    """
    A class used to generate a random password.

    Attributes:
        length (int): The length of the password to generate.
    """

    def __init__(self, length: int):
        """
        Initialize the PasswordGenerator object with a password length.

        Args:
            length (int): The desired length of the password.
        """
        self.length = length

    def generate_password(self) -> str:
        """
        Generate a random password with the specified length.

        The password contains lowercase letters, uppercase letters,
        numbers, and the exclamation mark character.

        Returns:
            str: A randomly generated password.
        """
        characters = (
            "abcdefghijklmnopqrstuvwxyz"
            "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
            "0123456789!"
        )

        password = ""

        for _ in range(self.length):
            password += random.choice(characters)

        return password


# Create an object
password_generator = PasswordGenerator(10)

# Call the main method
result = password_generator.generate_password()

# Print the result
print(result)
```

```python
class WordCounter:
    """
    A class used to count the occurrences of unique words in a list.

    Attributes:
        words (list[str]): A list of words to count.
    """

    def __init__(self, words: list[str]):
        """
        Initialize the WordCounter object with a list of words.

        Args:
            words (list[str]): The list of words to analyze.
        """
        self.words = words

    def count_words(self) -> str:
        """
        Count the occurrences of each unique word.

        Each word is converted to lowercase in the output, followed
        by the number of times it appears in the original list.

        Returns:
            str: A formatted string containing each word and its count.
        """
        unique_words = set(self.words)

        result = ""

        for word in unique_words:
            result += f"{word.lower()}:{self.words.count(word)}\n"

        return result


# Create an object
words = [
    "Welcome", "Ali", "Hi", "Ali", "No",
    "Hi", "No", "Ali", "No", "Ali"
]

word_counter = WordCounter(words)

# Call the main method
result = word_counter.count_words()

# Print the result
print(result)
```

```python
import random


class ChatBot:
    """
    A class used to create a simple rule-based chatbot.

    Attributes:
        responses (dict): A dictionary containing user messages
            and possible chatbot responses.
    """

    def __init__(self):
        """
        Initialize the ChatBot object with predefined responses.
        """
        self.responses = {
            "hello": ["Hello!", "Hi there!", "Greetings!"],
            "how are you": [
                "I'm doing well, thank you!",
                "I'm fine, how about you?"
            ],
            "goodbye": ["Goodbye!", "See you later!", "Farewell!"],
            "default": [
                "I'm sorry, I didn't understand.",
                "Could you please rephrase that?"
            ]
        }

    def get_response(self, user_input: str) -> str:
        """
        Get a random response based on the user's input.

        Args:
            user_input (str): The message entered by the user.

        Returns:
            str: A random response matching the user's input.
            If no match is found, a default response is returned.
        """
        for key in self.responses:
            if key in user_input:
                return random.choice(self.responses[key])

        return random.choice(self.responses["default"])

    def main_bot(self) -> None:
        """
        Run the chatbot and continuously receive user input.

        The chatbot responds to the user's messages and stops
        when the user enters 'goodbye'.

        Returns:
            None
        """
        print("Chatbot, hi! How can I help you?")

        while True:
            user_input = input("User: ").lower()

            response = self.get_response(user_input)

            print("Chatbot:", response)

            if user_input == "goodbye":
                break


# Create an object
chatbot = ChatBot()

# Call the main method
chatbot.main_bot()
```
