```python
def reverse_words(string: str) -> str:
    """
    Reverse the order of words in a given string.

    Words in the input are separated by exactly one space, with no
    leading or trailing spaces. The words themselves are not reversed,
    only their order in the string.

    :param string: The input string containing words separated by single spaces.
    :type string: str
    :return: A new string with the words in reverse order.
    :rtype: str
    """
    return " ".join(string.split(" ")[::-1])
```

```python
# 'battle no requires which that is victory greatest The'
reverse_words("The greatest victory is that which requires no battle")
```

```python
# 'hello'
reverse_words("hello")
```
