```python
def are_you_playing_banjo(name: str) -> str:
    """
    Determine whether a person is playing banjo based on their name.

    Rule: if the name starts with the letter "R" (uppercase or
    lowercase), the person is playing banjo.

    :param name: The name of the person to check.
    :type name: str
    :return: "<name> is playing banjo" if the name starts with 'R'/'r',
             otherwise "<name> is not playing banjo".
    :rtype: str
    """
    if name.startswith(("R", "r")):
        return f"{name} is playing banjo"
    return f"{name} is not playing banjo"
```

```python
are_you_playing_banjo("Rachel")
```

```python
are_you_playing_banjo("robert")
```

```python
are_you_playing_banjo("Sam")
```
