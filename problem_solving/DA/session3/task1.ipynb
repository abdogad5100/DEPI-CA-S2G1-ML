```python

def product_except_self(nums: list[int]) -> list[int]:
    """
    Return an array where each element is the product of all
    elements in nums except the element at the same index.

    Args:
        nums: A list of integers.

    Returns:
        A list containing the product of all elements except nums[i].
#     """
#     nums:
#        1    2    3    4
#        ↓    ↓    ↓    ↓

#     LEFT:
#        1    1    2    6

#     RIGHT:
#       24   12    4    1

    answer: list[int] = [1] * len(nums)

    # Calculate the product of elements to the left.
    prefix: int = 1

    for i in range(len(nums)):
        answer[i] = prefix
        prefix *= nums[i]

    # Calculate the product of elements to the right.
    suffix: int = 1

    for i in range(len(nums) - 1, -1, -1):
        answer[i] *= suffix
        suffix *= nums[i]

    return answer



```

```python
# Test Case 1
nums1: list[int] = [1, 2, 3, 4]
print(product_except_self(nums1))
# Expected: [24, 12, 8, 6]
```

```python
# Test Case 2
nums2: list[int] = [-1, 1, 0, -3, 3]
print(product_except_self(nums2))
# Expected: [0, 0, 9, 0, 0]
```
