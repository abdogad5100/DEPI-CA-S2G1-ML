```python
def calculate(s: str) -> int:
    """
    Evaluate a basic arithmetic expression containing +, -, *, and /.

    Args:
        s: A valid arithmetic expression as a string.

    Returns:
        The integer result of the expression.
    """
    stack: list[int] = []
    current_number: int = 0
    operation: str = "+"

    for i in range(len(s)):
        char: str = s[i]

        if char.isdigit():
            current_number = current_number * 10 + int(char)

        if char in "+-*/" or i == len(s) - 1:

            if operation == "+":
                stack.append(current_number)

            elif operation == "-":
                stack.append(-current_number)

            elif operation == "*":
                last_number: int = stack.pop()
                stack.append(last_number * current_number)

            elif operation == "/":
                last_number: int = stack.pop()
                stack.append(int(last_number / current_number))

            operation = char
            current_number = 0

    return sum(stack)
```

```python
s: str = "3+2*2"

result: int = calculate(s)

print(result)
```
