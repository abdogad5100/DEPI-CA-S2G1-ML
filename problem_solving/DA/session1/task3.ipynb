```python

def binary_search(arr: list[int], x: int) -> int:
    """
    Search for a target value in a sorted array using binary search.

    Args:
        arr: A sorted list of distinct integers.
        x: The target integer to search for.

    Returns:
        The index of x if it is found, otherwise -1.
    """

    low: int = 0
    high: int = len(arr) - 1

    while low <= high:
        mid: int = low + (high - low) // 2

        if arr[mid] == x:
            return mid

        elif arr[mid] < x:
            low = mid + 1

        else:
            high = mid - 1

    return -1


```

```python
# Test Case 1
arr1: list[int] = [1, 3, 5, 6]
x1: int = 5

print(binary_search(arr1, x1))
```

```python
# Test Case 2
arr2: list[int] = [1, 3, 5, 6]
x2: int = 2

print(binary_search(arr2, x2))
```
