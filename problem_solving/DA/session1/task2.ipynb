```python
def roman_to_int(s: str) -> int:
    """
    Convert a Roman numeral string into its integer value.

    Args:
        s: A string representing a valid Roman numeral.

    Returns:
        The integer value of the Roman numeral.
    """

    values: dict[str, int] = {
        "I": 1,
        "V": 5,
        "X": 10,
        "L": 50,
        "C": 100,
        "D": 500,
        "M": 1000,
    }

    result: int = 0

    for i in range(len(s)):
        current: int = values[s[i]]

        if i + 1 < len(s):
            next_value: int = values[s[i + 1]]

            if current < next_value:
                result -= current
            else:
                result += current

        else:
            result += current

    return result

```

```python
print(roman_to_int("III"))
```

```python
print(roman_to_int("IV"))
```
