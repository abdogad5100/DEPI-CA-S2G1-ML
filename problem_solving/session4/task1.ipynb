```python
def add_two_numbers(l1: list[int], l2: list[int]) -> list[int]:
    """
    Add two non-negative integers represented as lists of digits.

    Each list stores the digits of a number in reverse order (the
    first element is the least significant digit). The function adds
    the two numbers and returns the sum as a new list, also in
    reverse digit order.

    :param l1: Digits of the first number, least significant digit first.
    :type l1: list[int]
    :param l2: Digits of the second number, least significant digit first.
    :type l2: list[int]
    :return: Digits of the sum, least significant digit first.
    :rtype: list[int]
    """
    result = []
    carry = 0
    i = 0

    while i < len(l1) or i < len(l2) or carry:
        val1 = l1[i] if i < len(l1) else 0
        val2 = l2[i] if i < len(l2) else 0

        total = val1 + val2 + carry
        carry, digit = divmod(total, 10)

        result.append(digit)
        i += 1

    return result
```

```python
add_two_numbers([2, 4, 3], [5, 6, 4])
```

```python
add_two_numbers([0], [0])
```
